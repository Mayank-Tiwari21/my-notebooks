```python
from pyspark.sql import SparkSession
```

```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```python
import random
```

```python
spark = SparkSession.builder.appName("Learning the spark 2.0 ").getOrCreate()
print(spark)
```

```python
states = ['NY',"CAN","WS","FL","KL"]
items = ["item1","item2","item3","item3"]
```

```python
user_data = [(f"user_{id}",f"user_{id}@abc.com",random.choice(states)) for id in range(10000001)]
user_df = spark.createDataFrame(data = user_data,schema =["id","user_email","states"])
```

```python
user_df.show(n = 5)
```

```python
orders_data =  [(r,r,random.randint(0,9999),random.choice(states),10*r*0.2,random.choice(items)) for r in range(10000001)]
orders_df = spark.createDataFrame(data = orders_data,schema = ['transaction_id',"order_id","user_id","state","item"])
```
