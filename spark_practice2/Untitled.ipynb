```python
from pyspark.sql import SparkSession
```

```python
spark = SparkSession.builder.config("spark.sql.catalogImplementation", "hive").enableHiveSupport().appName("1Apr2025 practice").getOrCreate()
print(spark)
```

```python
data = [
    ("Alice", "USA", 25),
    ("Bob", "India", 30),
    ("Charlie", "USA", 35),
    ("David", "India", 40),
    ("Eve", "Canada", 28),
    ("Frank", "Canada", 38)
]

columns = ["name", "country", "age"]

df = spark.createDataFrame(data, columns)
```

```python
df.write \
    .mode("overwrite") \
    .partitionBy("country") \
    .format("parquet") \
    .save("file:///home/hdoop/notebooks/data/spark_practice2/1apr2025/")  
```

```python
df.write \
    .mode("overwrite") \
    .partitionBy("country") \
    .bucketBy(2, "age") \
    .format("parquet") \
    .saveAsTable("mytable")  # Works only for Hive tables
```
